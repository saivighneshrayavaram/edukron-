35.Write a function named generate_fibonacci().

```python
def generate_fibonacci(num):
    a=0
    b=1
    li=[0,1]
    for i in range(num-2):
        c=a+b
        li.append(c)
        a=b
        b=c
    return li
num=int(input("Enter number of terms:"))
d=generate_fibonacci(num)
print(d)
```
