40.Implement recursive category total for banking case 1; use sample reference `ACC-1040` and explain the result.

```python
Account_Number = "ACC-1040"
print("Account Number:", Account_Number)
accounts = [
    {"Account_Number": "ACC-1035", "Category": "Savings", "Balance": 50000},
    {"Account_Number": "ACC-1036", "Category": "Current", "Balance": 30000},
    {"Account_Number": "ACC-1037", "Category": "Savings", "Balance": 40000},
    {"Account_Number": "ACC-1040", "Category": "Savings", "Balance": 35000}
]
def total(accounts, category):
    if not accounts:
        return 0
    if accounts[0]["Category"] == category:
        return accounts[0]["Balance"] + total(accounts[1:], category)
    return total(accounts[1:], category)
result = total(accounts, "Savings")
print("Total Savings Balance:", result)
```

```python
Account_Number='ACC-1093'
print("Account Number:",Account_Number)
def transaction_limit(limit):
    def decorator(func):
        def wrapper(amount):
            if amount <= limit:
                return func(amount)
            else:
                return "Transaction exceeds the limit"
        return wrapper
    return decorator
@transaction_limit(50000)
def transfer(amount):
    return f"Transfer successful: ₹{amount}"
print(transfer(30000))
print(transfer(60000))
```
