1.Separate Positional and Keyword Arguments

```python
def display_arguments(*args,**kargs):
  print("Positional count:",len(args))
  print("Keyword count:",len(kargs))
  print("Positional values:",args)
  print("Keyword values:",kargs)
display_arguments(10,20,30,name="Rahul",city="Bangalore")
```

2.Sum Only Numeric Positional Arguments

```python
def calculate_numeric_total(*args):
  total=0
  for value in args:
    if(isinstance(value,(int,float))):
      total=total+value
  return total
result=calculate_numeric_total(10,"Pyhton",20.5,[1,2],30)
print("Numeric total:",result)
```

3.Find Largest and Smallest Number

```python
def find_min_max(*numbers):
  if(len(numbers)==0):
    return None,None
  else:
    largest=max(numbers)
    smallest=min(numbers)
    return largest,smallest
maximum,minimum=find_min_max(50,10,80,25,5)
print("Maximum:",maximum)
print("Minimum:",minimum)
```

4.Student Marks Report

```python
def student_report(name,*args,**kwargs):
  total=0;
  passed=all(mark >=35 for mark in args)
  print("Student:",name)
  print("Course:",kwargs.get("course","Not provided"))
  print("city:",kwargs.get("city","Not provided"))
  for value in args:
      total=total+value
  print("Total:",total)
  average=(total/len(args))
  print("Average:",average)
  if passed:
    print("Result: PASS")
  else:
    print("Result: FAIL")
student_report("Anil",80,75,90,65,course="Python",city="Hyderabad")
```

5.Employee Salary Calculator

```python
def calculate_salary(name,basic_salary,*args,**kargs):
  bonus=kargs.get("bonus",0)
  tax_percentage=kargs.get("tax",10)
  Gross_salary=basic_salary+sum(args)+bonus
  Tax_amount=Gross_salary*tax_percentage/100
  NetSalary=Gross_salary-Tax_amount
  print("Employee:",name)
  print("Gross Slary:",Gross_salary)
  print("Tax Amount:",Tax_amount)
  print("Net Salary:",NetSalary)
calculate_salary("Rahul", 50000, 5000, 3000, 2000, bonus=10000, tax=10, department="IT")
```

6.Count Data Types in *args

```python
def count_data_types(*args):
  integer_count=0
  string_count=0
  list_count=0
  other_count=0
  for value in args:
    if type(value) is int:
      integer_count += 1
    elif isinstance(value,str):
      string_count += 1
    elif isinstance(value,list):
      list_count += 1
    else:
      other_count += 1
  print("Integers:",integer_count)
  print("Strings:",string_count)
  print("Lists:",list_count)
  print("Others:",other_count)
count_data_types(10,"Python",[1,2],20,"SQL",3.5,True)
```

7.Filter Even and Odd Numbers

```python
def seperate_even_odd(*args):
  even_numbers=[]
  odd_numbers=[]
  for number in args:
    if(number%2==0):
      even_numbers.append(number)
    else:
      odd_numbers.append(number)
  return even_numbers,odd_numbers
evens,odds=seperate_even_odd(10,15,22,31,40,47)
print("Even numbers:",evens)
print("Odd umbers:",odds)
```

8.Search Employee Records

```python
def search_employees(*employees,**kargs):
  matched_employees=[]
  for employee in employees:
   matches=all(
      employee.get(key)==value
      for key,value in kargs.items()
   )
   if matches:
     matched_employees.append(employee)
  return matched_employees
employee1 = { "name": "Rahul", "department": "IT", "city": "Bangalore" }
employee2 = { "name": "Anil", "department": "HR", "city": "Hyderabad" }
employee3 = { "name": "Priya", "department": "IT", "city": "Bangalore" }
results=search_employees( employee1, employee2, employee3, department="IT", city="Bangalore" )
print(results)
```

9.Shopping Cart Total

```python
def checkout(*args,**kwargs):
  subtotal=sum(args)
  discount_percentage=kwargs.get("discount",0)
  tax_percentage=kwargs.get("tax",0)
  discount_amount=subtotal * discount_percentage/100
  discounted_amount=subtotal-discount_amount
  tax_amount=discounted_amount*tax_percentage/100
  final_amount=discounted_amount+tax_amount
  print("Customer:", kwargs.get("customer", "Guest"))
  print("Subtotal:", round(subtotal, 2))
  print("Discount:", round(discount_amount, 2))
  print("Tax:", round(tax_amount, 2))
  print("Final amount:", round(final_amount, 2))
checkout( 1000, 500, 250, customer="Rahul", discount=10, tax=18 )
```

11.Validate Required Keyword Arguments

```python
def register_user(**kwargs):
  missing_fields=[]
  required_fields=["name","email","password"]
  for field in required_fields:
    if field not in kwargs:
      missing_fields.append(field)
  if missing_fields:
    return "Missing Fields:",missing_fields
  else:
    print("Registration successful")
register_user( name="Rahul", email="rahul@example.com" )
```

12.Create a Dynamic SQL SELECT Query

```python
def build_select_query(table,*args,**kwargs):
  query=f"SELECT {','.join(args)} FROM {table}"
  if kwargs:
    conditions=[]
    for key,value in kwargs.items():
      conditions.append(f"{key}={value}")
    query +=" WHERE "+" AND ".join(conditions)
  return query
qquery= build_select_query( "Employee", "name", "salary", department="IT", city="Bangalore" )
print(qquery)
```

13.Function Forwarding with *args and **kwargs

```python
def execute_function(func,*args,**kwargs):
  print("Position Arguments:",args)
  print("Keyword arguments:",kwargs)
  result=func(args,kwargs)
  return result

def create_employee(name, age, department="IT", city="Bangalore"):
  cr={
      "name":name,
      "age":age,
      "department":department,
      "city":city
  }
  return cr
employee= execute_function( create_employee, "Rahul", 28, department="Data Engineering", city="Hyderabad" )
print(employee)
```

14.Decorator with *args and **kwargs

```python
def log_function_call(func):
 def wrapper(*args,**kwargs):
   print("Funcion name:",func.__name__)
   print("Positional:",args)
   print("Keyword:",kwargs)
   result=func(*args,**kwargs)
   print("Result: ",result)
 return wrapper
@log_function_call
def calculate_bill(price,quantity,discount=0):
  total=price*quantity
  discount_amount=total*discount/100
  return total-discount_amount
calculate_bill(500,3,discount=10)
```

15.Group Numbers Based on Conditions

```python
def classify_numbers(*args,**kwargs):
  lower_limit=kwargs.get("lower_limit",0)
  upper_limit=kwargs.get("upper_limit",100)
  below=[]
  within=[]
  above=[]
  for number in args:
    if number <lower_limit:
      below.append(number)
    elif number > upper_limit:
      above.append(number)
    else:
      within.append(number)
  return{
      "below":below,
      "within":within,
      "above":above
  }
classification = classify_numbers( 5, 15, 25, 35, 45, lower_limit=20, upper_limit=40 )
print(classification)
```

16.Merge Multiple Dictionaries

```python
def merge_dictionaries(*args,**kwargs):
  merged={}
  for dictionary in args:
    if(isinstance,dict):
      merged.update(dictionary)
  merged.update(kwargs)
  return merged
result = merge_dictionaries( {"name": "Rahul", "city": "Bangalore"}, {"department": "IT", "experience": 3}, {"experience": 4}, city="Hyderabad", salary=80000 )
print(result)
```

17.ETL Pipeline Summary

```python
import math
```

```python
def etl_summary(source,destination,*args,**kwargs):
  row_counts=kwargs.get("row_counts",{})
  batch_size=kwargs.get("batch_size",0)
  tables=args
  table_count=len(args)
  total_rows=sum(
      row_counts.get(table,0)
      for table in args
  )
  batches=math.ceil(total_rows/batch_size) if total_rows else 0
  print("Source:", source)
  print("Destination:", destination)
  print("Tables:", tables)
  print("Table count:", table_count)
  print("Total rows:", total_rows)
  print("Required batches:", batches)
etl_summary( "SQL Server", "ADLS Gen2", "Customer", "Orders", "Products", row_counts={ "Customer": 12000, "Orders": 55000, "Products": 3000 }, batch_size=10000 )
```

18.Role-Based Permission Checker

```python
def check_permissions(*args,**kwargs):
  role_permissions={"admin":{"read","write","update","delete"},
                    "editor":{"read","write","update"},
                    "viewer":{"read"}
                    }
  role=kwargs.get("role")
  allowed_actions=role_permissions.get(role)
  print("User:",kwargs.get("username","unknown"))
  print("Role:",role)
  for action in args:
    if action in allowed_actions:
      print(action,": ALLOWED")
    else:
      print(action,": DENIED")
check_permissions( "read", "update", "delete", username="Rahul", role="editor" )
```

19.Dynamic Object Creation

```python
class Employee:
  def __init__(self,*args,**kargs):
    self.skills=args
    for key,value in kargs.items():
      setattr(self,key,value)
  def display(self):
    print("Name:",getattr(self,"name","Not provided"))
    print("Department:",getattr(self,"department","Not provided"))
    print("Skills:",self.skills)
employee = Employee( "Python", "SQL", "Azure", name="Rahul", department="Data Engineering", experience=5 )
employee.display()
```

20.Complex Order Processor

```python
def process_order(order_id,*args,**kwargs):
  subtotal=0
  total_quantity=0
  for product in args:
    if not isinstance(product,dict):
      continue
    price=product.get("price",0)
    quantity=product.get("quantity",0)
    product_total=price*quantity
    subtotal +=product_total
    total_quantity += quantity
  discount_percentage=kwargs.get("discount",0)
  tax_percentage=kwargs.get("tax",0)
  discount_amount=subtotal*discount_percentage/100
  amount_after_discount=subtotal-discount_amount
  tax_amount=amount_after_discount*tax_percentage/100
  final_amount=amount_after_discount+tax_amount
  print("Order ID:", order_id)
  print("Customer:", kwargs.get("customer", "Guest"))
  print("Payment mode:", kwargs.get("payment_mode", "Not provided"))
  print("Total quantity:", total_quantity)
  print("Subtotal:", round(subtotal, 2))
  print("Discount amount:", round(discount_amount, 2))
  print("Tax amount:", round(tax_amount, 2))
  print("Final amount:", round(final_amount, 2))
process_order( "ORD-1001", { "name": "Laptop", "price": 50000, "quantity": 1 }, { "name": "Mouse", "price": 1000, "quantity": 2 }, { "name": "Keyboard", "price": 2000, "quantity": 1 }, customer="Rahul", payment_mode="UPI", discount=10, tax=18 )
```
